```python
import random
from unittest import result

def get_choices(): #### function
    options = ["rock", "paper", "scissors"]

    while True:
        player_choice = input("Select: rock, paper, or scissors: ").strip().lower()
        if player_choice in options:
            break
        else:
            print("Please enter rock, paper, or scissors")  #### User input


    computer_choice = random.choice(options)   #####random after importing random
    choices = {"player": player_choice , "computer": computer_choice} ##dictionary

    return choices


def check_win(player, computer):
        print(f"You chose: {player}, Computer chose: {computer}") ####fstring for concatenating string

        if player == computer:
            return "It's a tie!"

        elif player == "rock":
            if computer == "scissors":
                return "Rock smashes Scissors! You win!"
            else:
                return "Paper covers rock! You lose!" ####nesting

        elif player == "scissors":
            if computer == "paper":
                return "Scissors cuts paper! You win!"
            else:
                return "Rock smashes scissors! You lose!"

        elif player == "paper":
            if computer == "rock":
                return "Paper covers rock! You win!"
            else:
                return "Scissors cuts paper! You lose!"

choices = get_choices()
result = check_win(choices["player"], choices["computer"]) #####accessing dictionary
print(result)
```
